Train-Test Split 

```python
from sklearn.model_selection import train_test_split

X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_scaled,
    y_classification,
    test_size=0.2,
    random_state=42,
    stratify=y_classification
)

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_scaled,
    y_regression,
    test_size=0.2,
    random_state=42
)
```

AdaBoost Regression

```python
ada_regressor = AdaBoostRegressor(
    estimator=DecisionTreeRegressor(max_depth=4, random_state=42),
    n_estimators=100,
    learning_rate=1.0,
    random_state=42
)

ada_regressor.fit(X_train_reg, y_train_reg)

ada_reg_pred = ada_regressor.predict(X_test_reg)
```

Evaluate

```python
mae = mean_absolute_error(y_test_reg, ada_reg_pred)
mse = mean_squared_error(y_test_reg, ada_reg_pred)
rmse = np.sqrt(mse)

r2 = r2_score(y_test_reg, ada_reg_pred)

print("MAE :", mae)
print("MSE :", mse)
print("RMSE :", rmse)
print("R² :", r2)
```

Analyze how weak learners improve sequentially

```python
train_accuracy = []
test_accuracy = []

for y_pred_train, y_pred_test in zip(
    ada_classifier.staged_predict(X_train_cls), 
    ada_classifier.staged_predict(X_test_cls)
):
    train_accuracy.append(accuracy_score(y_train_cls, y_pred_train))
    test_accuracy.append(accuracy_score(y_test_cls, y_pred_test))
```

```python
plt.figure(figsize=(8, 5))

plt.plot(train_accuracy, label="Training Accuracy", color="blue")
plt.plot(test_accuracy, label="Testing Accuracy", color="orange")

plt.xlabel("Number of weak Learners")
plt.ylabel("Accuracy")

plt.title("AdaBoost: Improvement with sequential weak learners")

plt.legend()
plt.grid(True)

plt.show
```

# Part E - Voting and Stacking Ensemble

Implement Voting Classifier

Import libararies

```python
from sklearn.ensemble import VotingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
```

Hard Voting

```python
hard_voting = VotingClassifier(

    estimators=[
        ('lr', LogisticRegression(random_state=42)),
        ('dt', DecisionTreeClassifier(random_state=42)),
        ('knn', KNeighborsClassifier())
    ],

    voting='hard'
)

hard_voting.fit(X_train_cls, y_train_cls)

hard_pred = hard_voting.predict(X_test_cls)
```

Evaluation

```python
from sklearn.metrics import accuracy_score, classification_report

print("Hard Voting Accuracy:",
      accuracy_score(y_test_cls, hard_pred))

print()

print(classification_report(y_test_cls, hard_pred))
```

Compare Hard Voting vs Soft Voting

```python
soft_voting = VotingClassifier(

    estimators=[
        ('lr', LogisticRegression(random_state=42)),
        ('dt', DecisionTreeClassifier(random_state=42)),
        ('knn', KNeighborsClassifier())
    ],

    voting='soft'
)

soft_voting.fit(X_train_cls, y_train_cls)

soft_pred = soft_voting.predict(X_test_cls)
```

Evaluation

```python
print("Soft Voting Accuracy:",
      accuracy_score(y_test_cls, soft_pred))

print()

print(classification_report(y_test_cls, soft_pred))
```

Comparison Table

```python
comparison = pd.DataFrame({

    "Voting Method": [
        "Hard Voting",
        "Soft Voting"
    ],

    "Accuracy": [
        accuracy_score(y_test_cls, hard_pred),
        accuracy_score(y_test_cls, soft_pred)
    ]

})

comparison
```

Comparison Graph

```python
comparison.plot(

    x="Voting Method",

    y="Accuracy",

    kind="bar",

    legend=False,

    figsize=(6,4)

)

plt.title("Hard Voting vs Soft Voting")

plt.ylabel("Accuracy")

plt.show()
```

### Observation

Hard Voting selects the class predicted by the majority of classifiers, whereas Soft Voting averages the prediction probabilities before making a decision.

In most cases, Soft Voting performs slightly better because it considers the confidence level of each classifier instead of only their predicted labels.

Implement Stacking Classifier

Importing library

```python
from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression
```

```python
stack_classifier = StackingClassifier(

    estimators=[
        ('dt', DecisionTreeClassifier(random_state=42)),
        ('knn', KNeighborsClassifier()),
        ('lr', LogisticRegression(random_state=42))
    ],

    final_estimator=LogisticRegression(),

    cv=5

)

stack_classifier.fit(X_train_cls, y_train_cls)

stack_pred = stack_classifier.predict(X_test_cls)
```

Evaluation

```python
print("Stacking Accuracy:",
      accuracy_score(y_test_cls, stack_pred))

print()

print(classification_report(y_test_cls, stack_pred))
```

### Observation

The Stacking Classifier combines the predictions of multiple base models using a Logistic Regression Meta Learner. This often produces better performance than any individual base model because it learns how to combine their strengths.

# Part F - Model Evaluation and Comparison

Evaluate Classification Models

Importing libraries

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

classification_results = pd.DataFrame(columns=[
    "Model",
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "ROC-AUC"
])
```

Evaluate All Classification Models

```python
models = {

    "Decision Tree": dt_classifier,
    "Bagging": bag_classifier,
    "AdaBoost": ada_classifier,
    "Gradient Boosting": gb_classifier,
    "LightGBM": lgb_classifier,
    "XGBoost": xgb_classifier,
    "Voting": soft_voting,
    "Stacking": stack_classifier

}

for name, model in models.items():

    pred = model.predict(X_test_cls)

    if hasattr(model, "predict_proba"):
        prob = model.predict_proba(X_test_cls)[:,1]
        auc = roc_auc_score(y_test_cls, prob)
    else:
        auc = np.nan

    classification_results.loc[len(classification_results)] = [

        name,

        accuracy_score(y_test_cls,pred),

        precision_score(y_test_cls,pred),

        recall_score(y_test_cls,pred),

        f1_score(y_test_cls,pred),

        auc

    ]

classification_results
```

ROC Curve

```python
from sklearn.metrics import roc_curve

plt.figure(figsize=(8,6))

for name, model in models.items():

    if hasattr(model,"predict_proba"):

        prob = model.predict_proba(X_test_cls)[:,1]

        fpr,tpr,_ = roc_curve(y_test_cls,prob)

        plt.plot(fpr,tpr,label=name)

plt.plot([0,1],[0,1],'k--')

plt.xlabel("False Positive Rate")

plt.ylabel("True Positive Rate")

plt.title("ROC Curve Comparison")

plt.legend()

plt.show()
```

# Part G - Final Analysis and Reporting

## 1. Impact of Voting versus Boosting

### Voting Ensemble
- Voting combines predictions from multiple independent models.
- Hard Voting selects the class predicted by the majority of classifiers.
- Soft Voting averages prediction probabilities before making the final prediction.
- Voting is simple to implement and generally improves prediction stability.

### Boosting Ensemble
- Boosting trains models sequentially.
- Each new model focuses on correcting the mistakes made by previous models.
- Boosting usually achieves higher predictive accuracy than voting because it continuously reduces model errors.
- However, boosting requires more training time and computational resources.

### Analysis

In this project, boosting algorithms generally achieved better classification and regression performance than voting because they learned from previous prediction errors instead of treating all models equally.

## 2. Comparison of Tree-Based Boosting Algorithms

| Algorithm | Strengths | Limitations |
|------------|-----------|-------------|
| AdaBoost | Simple, reduces bias, easy to understand | Sensitive to noisy data and outliers |
| Gradient Boosting | High prediction accuracy, flexible | Slower training, may overfit without tuning |
| LightGBM | Fast training, low memory usage, efficient on large datasets | Requires parameter tuning for best performance |
| XGBoost | High accuracy, regularization, robust performance | Higher computational complexity than AdaBoost |

## 3. Advantages of Boosting and Stacking

### Advantages of Boosting
- Improves prediction accuracy.
- Reduces bias by learning from previous mistakes.
- Handles complex datasets effectively.
- Produces highly accurate predictive models.

### Advantages of Stacking
- Combines multiple different learning algorithms.
- Utilizes a Meta Learner to improve predictions.
- Often outperforms individual machine learning models.
- Provides better generalization on unseen data.

### Analysis

Boosting focuses on improving weak learners sequentially, whereas stacking combines diverse models using a Meta Learner. Both techniques significantly improve predictive performance compared to using a single machine learning model.
